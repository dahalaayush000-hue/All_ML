## Converters 

```python
def rename(name):
    if name== "Maharashtra Cricket Association Stadium":
        return "MIG"
    else:
        return name 
```

```python
rename("Maharashtra Cricket Association Stadium")
```

```python
pd.read_csv("../data/ipl_matches_data.csv", converters={"venue": rename})
```
