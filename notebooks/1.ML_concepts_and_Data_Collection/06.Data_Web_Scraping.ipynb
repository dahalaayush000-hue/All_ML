# Trying scrapping from Books to Scrape 

```python
import requests
from bs4 import BeautifulSoup

url= "https://books.toscrape.com/"
response= requests.get(url)
print(response.status_code)
print(response.text[:500])
```

```python
soup= BeautifulSoup(response.text,"html.parser") #We can use other parser like lxml.

print(soup.title.get_text(strip=True ))
```

```python
print(type(soup.title.get_text(strip=True )))
```

```python
books= soup.find_all("h3")
print(books)
```

```python
print(type(books))
```

```python
books_list=[]

for book in books:
    books_list.append(book.get_text(strip=True))  

print(books_list)
```

- The get_text function converts the html codes into clean python strings.
    - And the Strip function removes the unnecessary white spaces if available ..

```python
print(len(books_list))
```

## Now I a collecting the list from 50 pages ..

```python
import requests
from bs4 import BeautifulSoup

books_list= []

for page in range(1,51):
    url= f"https://books.toscrape.com/catalogue/page-{page}.html"

    response= requests.get(url)


    if(response.status_code==200):
        
        soup=BeautifulSoup(response.text,"html.parser")
        books= soup.find_all("h3")

        for book in books:
            books_list.append(book.get_text(strip=True))

        print(f"Page {page}: SUCCESS ({len(books)} items)")
    

    else:
         print(f"Page {page} Failed (Status:{response.status_code})")

print(len(books_list))
```

## That was for just the one column , list of the book  , now I want to add other features as well . So , I am goiing to do the same thing but by extracting the css class , that will be more easier..

```python
books_data = []

for page in range(1, 51):

    url = f"https://books.toscrape.com/catalogue/page-{page}.html"

    response = requests.get(url)
    soup = BeautifulSoup(response.text, "html.parser")

    books = soup.select("article.product_pod")

    for book in books:

        title = book.h3.a["title"] # .a helped to look the <a> attribute inside the h3 tag
        
        #.select_one means extracting the .classname from the css class .It returns the html element , so we need to get_text(strip)
        price = book.select_one(".price_color").get_text(strip=True)
        
        # Find the <p> tag with the "star-rating" class.
        # Then get its "class" attribute, which BeautifulSoup returns as a list,
        # such as ["star-rating", "Three"].
        # Finally, [1] selects the second item from that list, giving us the rating: "Three".
        rating = book.select_one("p.star-rating")["class"][1]   
        availability = book.select_one(".availability").get_text(strip=True)

        books_data.append({
            "title": title,
            "price": price,
            "rating": rating,
            "availability": availability
        })

    print(f"Page {page}: SUCCESS ({len(books)} items) | Total: {len(books_data)}")
```

```python
import pandas as pd

df = pd.DataFrame(books_data)

print(df.head())
```

```python
df.to_csv("(Web_Scrapped 2)books_dataset.csv", index=False)
```
