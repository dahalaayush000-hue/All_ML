# KNN Imputer 

- Euclidean Distance
- K is the number of neighbours
  

- the rows also may have the missing values so the euclidean distance is handled by the nan-euclidean distance formula .
- dist(x,y) = sqrt(weight * sq. distance from present coordinates) nd
-weight = Total # of coordinates / # of present coordinates

```python
import pandas as pd 
import numpy as np 

from sklearn.model_selection import train_test_split 
from sklearn.metrics import accuracy_score

from sklearn.impute import KNNImputer, SimpleImputer
from sklearn.linear_model import LogisticRegression
```

```python
df= pd.read_csv('../../../data/tested.csv')[['Age','Pclass','Fare','Survived']]
```

```python
df.head()
```

```python
df.isnull().mean()*100
```

```python
X = df.drop(columns=['Survived'])
y = df['Survived']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
X_train.head()
```

```python
#Weights default is uniform , and we can use the weights as distance also ..
knn=KNNImputer(n_neighbors=3,weights='distance')

X_train_trf=knn.fit_transform(X_train)
X_test_trf=knn.transform(X_test)
```

```python
new_df= pd.DataFrame(X_train_trf,columns=X_train.columns)
new_df['Age'].isnull().mean()*100 #  All filled so no missing values  here .
```

```python
lr=LogisticRegression()
lr.fit(X_train_trf,y_train)
y_pred=lr.predict(X_test_trf)
```

```python
accuracy_score(y_test,y_pred)
```

```python
# Comparision with Simple Imputer --> mean

si = SimpleImputer()

X_train_trf2 = si.fit_transform(X_train)
X_test_trf2 = si.transform(X_test)
```

```python
lr = LogisticRegression()

lr.fit(X_train_trf2,y_train)

y_pred2 = lr.predict(X_test_trf2)

accuracy_score(y_test,y_pred2)
```

### Almost the same here because of the small dataset but the KNN will outperform the mean median imputation,....
