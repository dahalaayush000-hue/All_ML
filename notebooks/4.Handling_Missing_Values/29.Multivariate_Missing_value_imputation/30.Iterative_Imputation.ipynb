# MICE (Multivariate Imputation by Chained Equations )

- Assumptions :
- MAR (Missing at random) -- Can predict the data by looking at the relationship between the other rows..
  

- Accurate but is slow and take memory as we have to upload all our training data into the server ..

- steps :
  ### Remove the target column

# Iteration 1
  ### Fill all the NAN values with the mean of the respctive columns

# Iteration 2-From left to right 

  ### Remove all Col1 missing values (NaN)
  - Yesma column 1 chai target column vayo missing values predict garna ko lagi ani 2 rw 3 chai input columns vayo whichever model use gare pani (1 ko missing lai) predict garna ko lagi ...
  ### Predict the missing vales using the other columns(2 and 3).
  

  ### Remove all the col2 missing values(NaN)
  ### Predict the missing values using the other columns(1 and 3).
- Yesma column 2 chai target column vayo missing values predict garna ko lagi ani 1 rw 3 chai input columns vayo whichever model use gare pani (2 ko missing lai)predict garna ko lagi ...

- 
  ### Remove all the col3 missing values(NaN)

  - Yesma column 3 chai target column vayo missing values predict garna ko lagi ani 2 rw 1 chai input columns vayo whichever model use gare pani (3 ko missing lai) predict garna ko lagi ...
  ### Predict the missing values using the other columns(1 and 2).

# Difference= iteration 2- Iteration1
- Yesle mean le fill gareko value rw iteration2(Model prediction) le fill gareko value ko difference find garxa .
- rw hamile yo steps lai difference zero or almost zero nahunjel samma predict garxam .. 

```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
```

```python
df = np.round(pd.read_csv('../../../data/50_Startups.csv')[['R&D Spend','Administration','Marketing Spend','Profit']]/10000)
np.random.seed(9)
df = df.sample(5)
df
```

```python
df = df.iloc[:,0:-1]
df
```

```python
df.iloc[1,0] = np.nan
df.iloc[3,1] = np.nan
df.iloc[-1,-1] = np.nan
```

```python
df.head()
```

```python
# Step 1 - Impute all missing values with mean of respective col

df0 = pd.DataFrame()

df0['R&D Spend'] = df['R&D Spend'].fillna(df['R&D Spend'].mean())
df0['Administration'] = df['Administration'].fillna(df['Administration'].mean())
df0['Marketing Spend'] = df['Marketing Spend'].fillna(df['Marketing Spend'].mean())
```

```python
# 0th Iteration
df0
```

```python
# Remove the col1 imputed value
df1 = df0.copy()

df1.iloc[1,0] = np.nan

df1
```

```python
# Use first 3 rows to build a model and use the last for prediction

X = df1.iloc[[0,2,3,4],1:3]
X
```

```python
y = df1.iloc[[0,2,3,4],0]
y
```

```python
lr = LinearRegression()
lr.fit(X,y)
lr.predict(df1.iloc[1,1:].values.reshape(1,2))
```

```python
df1.iloc[1,0] = 23.14
```

```python
df1
```

```python
# Remove the col2 imputed value

df1.iloc[3,1] = np.nan

df1
```

```python
# Use last 3 rows to build a model and use the first for prediction
X = df1.iloc[[0,1,2,4],[0,2]]
X
```

```python
y = df1.iloc[[0,1,2,4],1]
y
```

```python
lr = LinearRegression()
lr.fit(X,y)
lr.predict(df1.iloc[3,[0,2]].values.reshape(1,2))
```

```python
df1.iloc[3,1] = 11.06
```

```python
df1
```

```python
# Remove the col3 imputed value
df1.iloc[4,-1] = np.nan

df1
```

```python
# Use last 3 rows to build a model and use the first for prediction
X = df1.iloc[0:4,0:2]
X
```

```python
y = df1.iloc[0:4,-1]
y
```

```python
lr = LinearRegression()
lr.fit(X,y)
lr.predict(df1.iloc[4,0:2].values.reshape(1,2))
```

```python
df1.iloc[4,-1] = 31.56
```
