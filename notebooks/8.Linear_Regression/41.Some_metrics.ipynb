# 1.MAE(Mean Absolute Error)
- same unit
- robust to outliers

### Drawbacks 
- Graph of modulus is not differentiable at zero.

# 2.MSE (Mean Squared Error)

- can be use as loss funtion and is differentiable at all points.

### Drawbacks
- different units (squared units)
- Not robust to outliers.
  

# 3. RMSE (Root mean squared error)
- Same unit
- differentiable at all points

### Drawbacks
- Not that robust to outliers
