# Assumptions for Linear Regression

### 1.Linear Relationship between input and output column

### 2. NO Multicollinearity
- there shouldnt be any correlation between the input variables . They should be independent .
  

### 3. Normality of Residual (error)

### 4. Homoscedasticity
-Having the same type of spread or same scatter along the axis 

### 5. No Autocorrelation of error
