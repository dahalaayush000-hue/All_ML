```python
from sklearn.datasets import make_regression
import numpy as np
```

```python
X,y= make_regression(n_samples=4, n_features=1,n_targets=1,noise=80,random_state=13)
```

```python
import matplotlib.pyplot as plt
plt.scatter(X,y)
```

```python
# using ordinary Least squares Use linear regression
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()
lr.fit(X,y)
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
plt.scatter(X,y)
plt.plot(X,lr.predict(X),color="red")
```

# Applying Gradient Descent
- Assuming the slope constant 78.35, and randomly assuming the b value as 0 as a starting point to find the value of b where there will be the lowest loss function.

```python
y_pred=(78.35*X +0).reshape(4)
```

```python
plt.scatter(X,y)
plt.plot(X,lr.predict(X),color='red',label='OLS')
plt.plot(X,y_pred,color='blue',label='b=0')
plt.legend()
plt.show()
```

```python
m=78.35
b=0

loss_slope= -2* np.sum(y-m*X.ravel()-b)
loss_slope
```

```python
# If the learning rate =0.1
l_rate=0.1

step_size=l_rate* loss_slope
step_size
```

```python
#calculating the new intercept 
b=b-step_size
b
```

```python
y_pred1 = ((78.35 * X) + b).reshape(4)

plt.scatter(X,y)
plt.plot(X,lr.predict(X),color='red',label='OLS')
plt.plot(X,y_pred1,color='green',label='b = {}'.format(b))
plt.plot(X,y_pred,color='blue',label='b = 0')
plt.legend()
plt.show()
```

```python
loss_slope = -2 * np.sum(y - m*X.ravel() - b)
loss_slope
```

```python
b=b-step_size
b
```

```python

# Iteration 3
loss_slope = -2 * np.sum(y - m*X.ravel() - b)
loss_slope
```

```python

step_size = loss_slope*l_rate
step_size
```

```python
b = b - step_size
b
```

```python
y_pred2 = ((78.35 * X) + b).reshape(4)

plt.scatter(X,y)
plt.plot(X,lr.predict(X),color='red',label='OLS')
plt.plot(X,y_pred1,color='green',label='b = {}'.format(b))
plt.plot(X,y_pred2,color='black',label='b = {}'.format(b))
plt.plot(X,y_pred3,color='purple',label='b = {}'.format(b))
plt.plot(X,y_pred,color='blue',label='b = 0')
plt.legend()
plt.show()
```

```python
b = -100
m = 78.35
lr = 0.01

epochs = 100

for i in range(epochs):
  loss_slope = -2 * np.sum(y - m*X.ravel() - b)
  b = b - (lr * loss_slope)

  y_pred = m * X + b

  plt.plot(X,y_pred)

plt.scatter(X,y)
     
```
