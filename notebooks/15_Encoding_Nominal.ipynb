# OneHotEncoding(Nominal categorical Data)

- After dividing into different categories(making the dummy columns) , there may be the interdependance between the input columns , so one of the column is dropped(generally the first column) .. This is the concept of  multicollinearity..
- So if we divided into n categories, n-1 columns will be used .
