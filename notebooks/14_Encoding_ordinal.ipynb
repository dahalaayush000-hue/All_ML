- Ordinal Encoding and
- OneHotEncoding for nominal 

- Nominal(Where the order doesnt matter )
  ex:country , state , district 
  
- and ordinal categorical data (order matters)
 ex: rank(1,2,3), or review(excellent, good, average, bad)

# ordinal encoding 
- If we want to encode the input columns we use the ordinal encoding .. but
- If the output column is to be encoded then the label encoding is used .
